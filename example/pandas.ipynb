## 資料的三個入口：API / CSV / 資料庫

前面都是用 list of dict 建 df（API 回傳的形狀）。另外兩個常見入口：`read_csv` 讀檔案、`read_sql` 讀資料庫。

```python
# 入口②：CSV 檔 → DataFrame（第 2 章存下的輸出檔）
df_csv = pd.read_csv("../output/TaiwanStockPrice_2330.csv")
print(df_csv.shape)   # (349, 10)
df_csv.head(3)
```

```python
# 入口③：資料庫 → DataFrame（把第 5 章 to_sql 寫進去的表讀回來）
# to_sql 和 read_sql 是一對：一個把 df 寫進 MySQL、一個把表讀回成 df
from sqlalchemy import create_engine, text
engine = create_engine("mysql+pymysql://root:1234@127.0.0.1:3306/mydb")

df_db = pd.read_sql("SELECT * FROM TaiwanStockPrice", engine)
print(df_db.shape)   # (1740, 10)
df_db.dtypes         # 注意 stock_id 的型別——由資料庫欄位型別決定
```

## df 版去重寫入（對照手冊06 的 upsert）

手冊06 的主線是「DB 層去重」：主鍵擋重複、upsert 撞到改更新。
同一件事也可以完全在 pandas 裡做，流程六步：

1. `read_sql` 把庫裡現有的資料讀回來
2. `concat` 新舊合併
3. 統一主鍵欄位型別（型別不一致時 `drop_duplicates` 會視為不同筆，去重不會發生）
4. `drop_duplicates(subset=[主鍵欄], keep="last")` 去重——同股同日留最新
5. `DELETE FROM` 清空表（表結構和主鍵保留）
6. `to_sql(if_exists="append")` 整批寫回

`drop_duplicates` 就是「去重」的 DataFrame 版本：`subset` 對應複合主鍵、`keep="last"` 對應 upsert 的「更新成新值」。

取捨：DB 層是增量更新、多 worker 安全，適合正式排程；df 層整表重寫，適合小資料、一次性整理與教學。

```python
# 建一張帶複合主鍵的實驗表（跟手冊06 的表同構）
with engine.begin() as conn:
    conn.execute(text("DROP TABLE IF EXISTS TaiwanStockPrice_dfdemo"))
    conn.execute(text("""CREATE TABLE TaiwanStockPrice_dfdemo (
      date DATE NOT NULL, stock_id VARCHAR(50) NOT NULL,
      Trading_Volume BIGINT, Trading_money BIGINT,
      open FLOAT, max FLOAT, min FLOAT, close FLOAT,
      spread FLOAT, Trading_turnover BIGINT,
      PRIMARY KEY (stock_id, date))"""))

# 模擬「這次抓回來的一批」：取 2330 的資料
new_batch = df_db[df_db["stock_id"].astype(str) == "2330"].copy()
len(new_batch)   # 349
```

```python
def df_upsert(new_df):
    current = pd.read_sql("SELECT * FROM TaiwanStockPrice_dfdemo", engine)   # ① 讀回現有資料
    merged = pd.concat([current, new_df], ignore_index=True)                 # ② 新舊合併
    merged["stock_id"] = merged["stock_id"].astype(str)                      # ③ 統一主鍵欄位型別
    merged["date"] = pd.to_datetime(merged["date"])
    merged = merged.drop_duplicates(subset=["stock_id", "date"], keep="last")  # ④ 去重
    with engine.begin() as conn:
        conn.execute(text("DELETE FROM TaiwanStockPrice_dfdemo"))            # ⑤ 清空表
        merged.to_sql("TaiwanStockPrice_dfdemo", conn, if_exists="append", index=False)  # ⑥ 整批寫回

def count():
    return pd.read_sql("SELECT COUNT(*) c FROM TaiwanStockPrice_dfdemo", engine)["c"][0]

df_upsert(new_batch)
print("第一輪：", count())   # 349
df_upsert(new_batch)
print("第二輪：", count())   # 349——重跑筆數不變，這就是冪等
```

```python
# 改一筆值再跑：筆數不變、值被更新（對應 upsert 的 update 那一半）
new_batch.loc[new_batch.index[0], "close"] = 999.0
df_upsert(new_batch)
d0 = str(new_batch.iloc[0]["date"])[:10]
v = pd.read_sql(text("SELECT close FROM TaiwanStockPrice_dfdemo WHERE stock_id=:s AND date=:d"),
                engine, params={"s": "2330", "d": d0})
print("筆數：", count(), "| close 更新為：", v["close"][0])   # 349 | 999.0
```

```python
# 收尾：刪掉實驗表，資料庫回到原狀
with engine.begin() as conn:
    conn.execute(text("DROP TABLE TaiwanStockPrice_dfdemo"))
print("done")
```
